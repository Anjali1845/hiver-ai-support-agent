```python
import pandas as pd

# Load the Twitter dataset
df = pd.read_csv("../data/raw/twcs.csv")

print("Dataset shape:", df.shape)
df.head()     
```

```python
# See all column names
print(df.columns.tolist())
```

```python
# Find all support brands
brands = df[df["inbound"] == False]["author_id"].value_counts()

print("Number of brands:", len(brands))
brands.head(20)
```

```python
# Filter only Spotify support conversations

spotify = df[
    (df["author_id"] == "SpotifyCares") |
    (df["text"].str.contains("@SpotifyCares", case=False, na=False))
]

print("Spotify dataset shape:", spotify.shape)

spotify.head()
```

```python
# Count customer vs support tweets

spotify["inbound"].value_counts()
```

```python
# Save cleaned Spotify dataset

spotify.to_csv("../data/processed/spotify_tweets.csv", index=False)

print("Saved successfully!")
```
