```python
import pandas as pd

spotify = pd.read_csv("../data/processed/spotify_tweets.csv")

print(spotify.shape)
spotify.head()
```

```python
customers = spotify[spotify["inbound"] == True].copy()

print("Customer tweets:", customers.shape)

customers.head()
```

```python
sample = customers.sample(n=200, random_state=42)

sample = sample[["tweet_id", "text"]]

sample.head()
```

```python
sample.to_csv("../data/golden_set.csv", index=False)

print("Golden set created!")
```
